```python
# 11. Find total sales.
total_sales=df["sales"].sum()
# 12. Find total quantity sold. 
total_quantity=df["quantity"].sum()
# 13. Find total number of orders.
total_orders=df["order_id"].count() 
# 14. Find average order value.
avg_order_value= df["sales"].mean()
# 15. Find average quantity per order.
avg_quantity_order=df["quantity"].mean()
# 16. Find the highest sales order.
hightest_sales_order=df["sales"].max()
# 17. Find the lowest sales order.
lowest_sales_order=df["sales"].min() 
# 18. Find the average unit price.
avg_unit_price=df["unit_price"].mean()
```

```python
# 48. Calculate sales growth between months.
monthly_sales = df.groupby(df["order_date"].dt.to_period("M"))["sales"].sum()
```

```python
sales_growth=monthly_sales.pct_change()*100
```

```python
sales_growth
```

```python
sales_growth.plot(kind="line")
plt.xlabel("months")
plt.ylabel("sales")
plt.title("sales growth")
plt.show()
```
